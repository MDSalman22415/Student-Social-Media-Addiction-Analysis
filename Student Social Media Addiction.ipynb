# Which age group the highest sleeping hours

```python
df['Avg_Daily_Usage_Hours'] = pd.cut(
    df['Avg_Daily_Usage_Hours'],
    bins=[1.0,2.0,5.0,8.0,10.0],
    labels=['Minimum Sleep','Low Sleep','Medium Sleep','High Sleep']
)
```

```python
df.head()
```

```python
highest_sleep_hour = df.groupby('Avg_Daily_Usage_Hours')['Age'].mean().reset_index()
```

```python
highest_sleep_hour
```

# Visualization

```python
df.head()
```

# Which age group the highest sleeping hours


```python
age_group = df.groupby('Avg_Daily_Usage_Hours')['Age'].mean().reset_index()
```

```python
age_group
```

```python
fig = px.bar(age_group,
            x='Avg_Daily_Usage_Hours',
            y='Age',
            color='Avg_Daily_Usage_Hours',
            title='Highest sleeping hours by age group',
            labels={'Avg_Daily_Usage_Hours':'Hours','Age':'Age'})

fig.show()
```

# What is the Addicted_Score of Academic_Level

```python
addicted_score = df.groupby('Academic_Level')['Addicted_Score'].mean().reset_index()
```

```python
addicted_score
```

```python
fig = px.pie(addicted_score,
            names='Academic_Level',
            values='Addicted_Score',
            hole=0.3,
            color='Academic_Level',
            title='Academic_Level vs Addicted Score',
            color_discrete_sequence=px.colors.qualitative.Pastel)
fig.show()
```
